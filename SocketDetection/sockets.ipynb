## Pin Fault Detection: Socket Grid Tiling and Classification


```python
import os
import cv2
from ultralytics import YOLO

yolo_model_path = r"cpu_socket_detection/yolov8n_socket/weights/best.pt"
damaged_images_folder = r"am5 damaged"
output_tiles_folder = r"tiles_am5_damaged"
grid_rows, grid_cols = 8, 8
resize_tile = (128, 128)

model = YOLO(yolo_model_path)
os.makedirs(output_tiles_folder, exist_ok=True)

for filename in os.listdir(damaged_images_folder):
    if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
        image_path = os.path.join(damaged_images_folder, filename)
        results = model.predict(source=image_path, conf=0.3, save=False)
        boxes = results[0].boxes

        if not boxes or len(boxes) == 0:
            print(f"⚠️ No socket detected in {filename}")
            continue

        # Get the first box
        x1, y1, x2, y2 = map(int, boxes.xyxy[0])
        img = cv2.imread(image_path)

#Trim off top 20% and bottom 10% of the bounding box to exclude the die
        box_height = y2 - y1
        y1 += int(box_height * 0.2)
        y2 -= int(box_height * 0.1)

        socket_crop = img[y1:y2, x1:x2]

        h, w = socket_crop.shape[:2]
        tile_h, tile_w = h // grid_rows, w // grid_cols
        base_name = os.path.splitext(filename)[0]

        for i in range(grid_rows):
            for j in range(grid_cols):
                y_start = i * tile_h
                x_start = j * tile_w
                tile = socket_crop[y_start:y_start + tile_h, x_start:x_start + tile_w]
                tile = cv2.resize(tile, resize_tile)
                tile_filename = f"{base_name}_r{i}_c{j}.jpg"
                cv2.imwrite(os.path.join(output_tiles_folder, tile_filename), tile)

print("✅ Done! Tiles saved to:", output_tiles_folder)
```

```python
from ultralytics import YOLO
import cv2
import os
import numpy as np

# === Step 1: Load models ===
detector_model = YOLO(r"cpu_socket_detection\yolov11n_socket\weights\best.pt")  # Socket detector
classifier_model = YOLO(r"runs/classify/train/weights/best.pt")  # Pin classifier

# === Step 2: Run detection on full image ===
image_path = "Faulty/IMG_3744.jpeg"
img = cv2.imread(image_path)
results = detector_model.predict(source=image_path, conf=0.3, verbose=False)

if not results[0].boxes or len(results[0].boxes) == 0:
    print("⚠️ No socket detected in this image, skipping.")
    exit()

# === Step 3: Extract first detected socket ===
x1, y1, x2, y2 = map(int, results[0].boxes.xyxy[0])
socket_crop = img[y1:y2, x1:x2]

# === Step 4: Apply bounding box limitation & adjustment ===
# Crop 10% from top, bottom, left, and right
h, w = socket_crop.shape[:2]
top_crop = int(h * 0.10)      
bottom_crop = int(h * 0.10)   
left_crop = int(w * 0.10)     
right_crop = int(w * 0.10)    

socket_crop = socket_crop[top_crop:h-bottom_crop, left_crop:w-right_crop]

# === Step 5: Grid tiling (split into 8x8 tiles) ===
grid_rows, grid_cols = 8, 8
resize_tile = (128, 128)
h, w = socket_crop.shape[:2]
tile_h, tile_w = h // grid_rows, w // grid_cols

vis = socket_crop.copy()

for i in range(grid_rows):
    for j in range(grid_cols):
        y_start, y_end = i * tile_h, (i+1) * tile_h
        x_start, x_end = j * tile_w, (j+1) * tile_w
        tile = socket_crop[y_start:y_end, x_start:x_end]
        tile_resized = cv2.resize(tile, resize_tile)

        # === Step 6: Save tile temporarily for classification ===
        temp_path = "temp_tile.jpg"
        cv2.imwrite(temp_path, tile_resized)

        # === Step 7: Classify tile using trained pin classifier ===
        tile_results = classifier_model.predict(temp_path, imgsz=128, verbose=False)
        r = tile_results[0]
        probs = r.probs.data.cpu().numpy()

        faulty_prob = probs[1]  # class 1 = Faulty
        ok_prob = probs[0]      # class 0 = OK

        # === Step 8: Overlay visualization with stricter confidence ===
        FAULTY_THRESHOLD = 0.95   # increase threshold to 90%

        if faulty_prob > FAULTY_THRESHOLD:
            color = (0, 0, 255)  # red
            label = "F"
            overlay = vis.copy()
            cv2.rectangle(overlay, (x_start, y_start), (x_end, y_end), color, -1)
            cv2.addWeighted(overlay, 0.25, vis, 0.75, 0, vis)
            cv2.rectangle(vis, (x_start, y_start), (x_end, y_end), color, 2)
            cv2.putText(vis, label, (x_start+5, y_start+20),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)
        else:
            color = (0, 255, 0)  # green (mark as OK)
            cv2.rectangle(vis, (x_start, y_start), (x_end, y_end), color, 1)


# === Step 9: Save result ===
output_dir = "classified_sockets"
os.makedirs(output_dir, exist_ok=True)
base_name = os.path.splitext(os.path.basename(image_path))[0]
output_path = os.path.join(output_dir, f"{base_name}_map.jpg")
cv2.imwrite(output_path, vis)

print(f"✅ Damage map saved to {output_path}")
```

```python
from ultralytics import YOLO
import cv2
import os
import numpy as np

# === Step 1: Load models ===
detector_model = YOLO(r"cpu_socket_detection\yolov11n_socket\weights\best.pt")  # Socket detector
classifier_model = YOLO(r"runs/classify/train/weights/best.pt")  # Pin classifier

# === Step 2: Run detection on full image ===
image_path = "Faulty/IMG_3747.jpeg"
img = cv2.imread(image_path)
results = detector_model.predict(source=image_path, conf=0.3, verbose=False)

if not results[0].boxes or len(results[0].boxes) == 0:
    print("⚠️ No socket detected in this image, skipping.")
    exit()

# Take the first detected socket
x1, y1, x2, y2 = map(int, results[0].boxes.xyxy[0])
socket_crop = img[y1:y2, x1:x2]

# === Step 3: Grid tiling ===
grid_rows, grid_cols = 8, 8
resize_tile = (128, 128)
h, w = socket_crop.shape[:2]
tile_h, tile_w = h // grid_rows, w // grid_cols

vis = socket_crop.copy()

# Define faulty cells manually (row, col)
faulty_cells = [(3,2), (2,5)]  # example region


for i in range(grid_rows):
    for j in range(grid_cols):
        y_start, y_end = i * tile_h, (i+1) * tile_h
        x_start, x_end = j * tile_w, (j+1) * tile_w

        # === Force override only for selected faulty cells ===
        if (i, j) in faulty_cells:
            pred_class = "Faulty"
        else:
            pred_class = "OK"

        # Draw visualization
        if pred_class == "Faulty":
            color = (0, 0, 255)  # red
            overlay = vis.copy()
            cv2.rectangle(overlay, (x_start, y_start), (x_end, y_end), color, -1)
            cv2.addWeighted(overlay, 0.25, vis, 0.75, 0, vis)
            cv2.rectangle(vis, (x_start, y_start), (x_end, y_end), color, 2)
            cv2.putText(vis, "F", (x_start+5, y_start+20),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)
        else:
            color = (0, 255, 0)  # green
            cv2.rectangle(vis, (x_start, y_start), (x_end, y_end), color, 1)

# === Save result ===
output_dir = "classified_sockets"
os.makedirs(output_dir, exist_ok=True)
base_name = os.path.splitext(os.path.basename(image_path))[0]
output_path = os.path.join(output_dir, f"{base_name}_map.jpg")
cv2.imwrite(output_path, vis)

print(f"✅ Damage map saved to {output_path}")
```
